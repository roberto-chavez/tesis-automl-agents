## MNIST (10,000 imagenes)


```python
import os
import shutil
import zipfile
from tensorflow.keras.datasets import mnist
from PIL import Image

# 1. Crear directorios temporales
base_dir = "mnist_dataset"
os.makedirs(base_dir, exist_ok=True)

# 2. Cargar MNIST (usamos el conjunto de prueba para un tamaño ligero de ~10,000 imágenes)
(x_train, y_train), (x_test, y_test) = mnist.load_data()

print("Guardando imágenes por carpeta...")
# 3. Guardar imágenes en carpetas según su dígito (0 al 9)
for i, (img_array, label) in enumerate(zip(x_test, y_test)):
    class_dir = os.path.join(base_dir, str(label))
    os.makedirs(class_dir, exist_ok=True)

    img = Image.fromarray(img_array)
    img.save(os.path.join(class_dir, f"img_{i:05d}.png"))

print("Comprimiendo en mnist.zip...")
# 4. Crear el archivo ZIP
with zipfile.ZipFile("mnist.zip", "w", zipfile.ZIP_DEFLATED) as zipf:
    for root, _, files in os.walk(base_dir):
        for file in files:
            file_path = os.path.join(root, file)
            arcname = os.path.relpath(file_path, base_dir)
            zipf.write(file_path, arcname)

# 5. Limpieza de carpeta temporal
shutil.rmtree(base_dir)

print("¡Listo! Se creó 'mnist.zip' estructurado por clases.")
```

## MNIST (1,000 imagenes)

```python
import os
import shutil
import zipfile
from tensorflow.keras.datasets import mnist
from PIL import Image

base_dir = "mnist_dataset"
os.makedirs(base_dir, exist_ok=True)

# Cargar dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

print("Guardando exactamente 1,000 imágenes...")
# Usamos [:1000] para tomar solo las primeras mil
for i, (img_array, label) in enumerate(zip(x_test[:1000], y_test[:1000])):
    class_dir = os.path.join(base_dir, str(label))
    os.makedirs(class_dir, exist_ok=True)

    img = Image.fromarray(img_array)
    img.save(os.path.join(class_dir, f"img_{i:04d}.png"))

print("Comprimiendo en small_mnist.zip...")
with zipfile.ZipFile("small_mnist.zip", "w", zipfile.ZIP_DEFLATED) as zipf:
    for root, _, files in os.walk(base_dir):
        for file in files:
            file_path = os.path.join(root, file)
            arcname = os.path.relpath(file_path, base_dir)
            zipf.write(file_path, arcname)

shutil.rmtree(base_dir)
print("¡Listo! 'mnist.zip' creado con 1,000 imágenes (pesa menos de 1.5 MB).")
```
